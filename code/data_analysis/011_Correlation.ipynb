```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import linregress
```

```python
df= pd.read_csv(r"C:\Users\LENOVO\Documents\PostGrad\BSc Hons\4ZOL509_Research\data\abundance_biomass.csv")
```

```python
# Compute correlation matrix (drop non-numerical columns)
corr_matrix = df.drop(columns=['prawn_id', 'replicate', 'site', 'sample']).corr(numeric_only=True)

# Create a mask for the upper triangle
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# Set up plot grid
fig = plt.figure()

# Plot 4: Correlation Heatmap – Drop non-numerical columns
corr_matrix = df.drop(columns=['prawn_id', 'replicate', 'site', 'sample']).corr(numeric_only=True)
sns.heatmap(corr_matrix, mask=mask, annot=True, cmap='coolwarm', square=True, cbar=True)
plt.title('Correlation Matrix')

# Adjust layout for better spacing
plt.tight_layout()
plt.show()
```
