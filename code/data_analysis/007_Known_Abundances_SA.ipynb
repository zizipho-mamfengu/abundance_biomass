```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path

plot_path = Path("fig")
plot_path.mkdir(exist_ok=True)
# Data
estuaries = [
    "Great Brak (WC)",
    "Langebaan (WC)",
    "Swartkops (EC)",
    "Kasouga (EC)",
    "Durban Bay (KZN)"
]

# maximum abundances
max_abundance = [156, 419, 600, 400, 264]

x = np.arange(len(estuaries))

# Plot
fig, ax = plt.subplots(dpi=500)

# Plot abundance as bars
for i in range(len(estuaries)):
    ax.bar(estuaries[i], max_abundance[i], color="skyblue")
    ax.text(i, max_abundance[i] * 1.0, str(max_abundance[i]), 
            ha="center", va="bottom")

# Labels
ax.set_xlabel("Estuaries")
ax.set_ylabel("Abundance (individuals·m⁻²)")
ax.set_title(r"Summary of Previous Findings on $\it{Kraussillichirus\ kraussi}$")

plt.xticks(rotation=30, ha="right")
plt.tight_layout()

plt.tight_layout()
plt.savefig(plot_path / "known_Abundance", dpi=300)
plt.show()
```
