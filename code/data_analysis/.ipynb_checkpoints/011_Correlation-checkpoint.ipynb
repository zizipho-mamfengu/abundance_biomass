```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import linregress
```

```python
df= pd.read_csv(r"C:\Users\LENOVO\Documents\PostGrad\BSc Hons\Research\data\abundance_biomass.csv")
```

```python
# Set up the 2x2 subplot grid
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Plot 1: Scatter plot – Carapace vs Total Length by Site
sns.scatterplot(data=df, x='total_length', y='carapace', hue='site', ax=axes[0, 0])
axes[0, 0].set_title('Carapace vs Total Length by Site')

# Plot 2: Boxplot – Total Length by Site
sns.boxplot(data=df, x='site', y='total_length', ax=axes[0, 1])
axes[0, 1].set_title('Total Length of Prawns Across Sites')

# Plot 3: Histogram – Egg Count Distribution
sns.histplot(df['eggs'], bins=15, kde=True, ax=axes[1, 0])
axes[1, 0].set_title('Egg Count Distribution')

# Plot 4: Correlation Heatmap – Drop non-numerical columns
corr_matrix = df.drop(columns=['prawn_id', 'replicate', 'site', 'sample']).corr(numeric_only=True)
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', ax=axes[1, 1])
axes[1, 1].set_title('Correlation Matrix')

# Adjust layout for better spacing
plt.tight_layout()
plt.show()
```
