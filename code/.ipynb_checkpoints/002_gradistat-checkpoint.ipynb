```python
# Importing the pandas library for data manipulation and analysis
import pandas as pd
# Importing the numpy library for numerical operations
import numpy as np
# Suppressing warning messages (useful for cleaner outputs during analysis)
import warnings
warnings.filterwarnings('ignore')  # This line disables warning messages that might pop up from pandas/numpy, etc.
```

```python
# Reading a CSV file from the specified path and loading it into a pandas DataFrame called 'df'
# 'r' before the string indicates a raw string, which treats backslashes as literal characters (avoiding escape sequences)
df = pd.read_csv(r"C:\Users\LENOVO\Documents\PostGrad\BSc Hons\Research\seive_percentage.csv")
```

```python
# Define sieve sizes in microns, ordered from coarse to fine
sieve_sizes_microns = np.array([2000, 1000, 500, 250, 125, 63, 1])

# Convert sieve sizes to phi units using the formula: phi = -log2(diameter in mm)
# Dividing by 1000 converts microns to mm before taking the log base 2
phi_sizes = -np.log2(sieve_sizes_microns / 1000.0)

sieve_columns = ["sieve_2000um", "sieve_1000um", "sieve_500um", "sieve_250um", "sieve_125um", "sieve_63um", "sieve_1um"]

# Define a function to compute Gradistat sediment statistics based on Folk & Ward (1957)
def compute_gradistat(row):
    # Define the list of column names in df_perc that contain the percentage data
    # Extract the percentage weights for each sieve from the current row
    freq = row[sieve_columns].values
    
    # Calculate cumulative frequency (%), normalized so that the final total is 100%
    cum_freq = np.cumsum(freq) / np.sum(freq) * 100

    # Define percentiles to interpolate (standard in sedimentology)
    percentiles = [5, 16, 25, 50, 75, 84, 95]

    # Interpolate phi values at the specified percentiles
    phi_interp = np.interp(percentiles, cum_freq, phi_sizes)

    # Unpack the interpolated values
    phi5, phi16, phi25, phi50, phi75, phi84, phi95 = phi_interp

    # Calculate Folk & Ward statistical parameters:
    
    # Mean grain size (phi units): average of 16th, 50th, and 84th percentiles
    mean_phi = (phi16 + phi50 + phi84) / 3

    # Sorting (phi units): reflects sediment uniformity
    sorting = ((phi84 - phi16) / 4) + ((phi95 - phi5) / 6.6)

    # Skewness: measure of asymmetry in grain size distribution
    skewness = (((phi16 + phi84 - 2 * phi50) / (2 * (phi84 - phi16))) +
                ((phi5 + phi95 - 2 * phi50) / (2 * (phi95 - phi5))))

    # Kurtosis: measure of peakedness in the distribution
    kurtosis = (phi95 - phi5) / (2.44 * (phi75 - phi25))

    # Return a pandas Series with calculated values, which can be added to a DataFrame
    return pd.Series({
        "mean_phi": mean_phi,
        "sorting": sorting,
        "skewness": skewness,
        "kurtosis": kurtosis
    })
```

```python
# Apply function
results = df.groupby(["site", "replicates"]).apply(compute_gradistat).reset_index()
```

```python
# Save the 'df_perc' DataFrame to a CSV file named 'seive_percentage.csv'
# index=False ensures that the row index is not written to the CSV file
results.to_csv('gradistat.csv', index=False)

# Print a message to confirm that the file was saved successfully
print("CSV file saved successfully!")
```
