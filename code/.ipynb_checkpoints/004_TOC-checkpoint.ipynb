```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
toc=pd.read_csv(r"C:\Users\LENOVO\Documents\PostGrad\BSc Hons\4ZOL509_Research\data\toc.csv")
```

```python
# Step 2: Calculate organic matter lost (this is second step)
toc["organic_matter"] = toc["dry_weight"] - toc["ash_weight"]
toc["toc_percent"] = (toc["organic_matter"] / toc["dry_weight"]) * 100 * 0.58
```

```python

# Step 3: Group by site to get mean and standard deviation
grouped = toc.groupby('site')['toc_percent'].agg(['mean', 'std']).reset_index()

# Step 4: Plot the TOC% per site
plt.figure(figsize=(8, 6))
plt.bar(grouped['site'], grouped['mean'], yerr=grouped['std'], capsize=5, color=['#6495ED', '#90EE90', '#FFA07A'])
plt.ylabel('TOC (%)')
plt.title('Total Organic Content (TOC) by Site')
plt.ylim(0, max(grouped['mean'] + grouped['std']) + 1)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()
```
