```python
# Importing the pandas library for data manipulation and analysis
import pandas as pd
# Suppressing warning messages (useful for cleaner outputs during analysis)
import warnings
warnings.filterwarnings('ignore')  # This line disables warning messages that might pop up from pandas/numpy, etc.
```

```python
# Reading a CSV file from the specified path and loading it into a pandas DataFrame called 'df'
# 'r' before the string indicates a raw string, which treats backslashes as literal characters (avoiding escape sequences)
df = pd.read_csv(r"C:\Users\LENOVO\Documents\PostGrad\BSc Hons\Research\may_sandprawns.csv")
```

```python
# Creating an empty DataFrame to store percentage values
df_perc = pd.DataFrame()
# Initializing the new DataFrame 'df_perc' with only the 'site' and 'replicates' columns from the original DataFrame
df_perc = df[['site', 'replicates']]
# Defining a list of column names corresponding to the sediment retained on each sieve
sieve_columns = [
    'sieve_2um', 'sieve_1um', 'sieve_500um',
    'sieve_250um', 'sieve_125um', 'sieve_63um', 'sieve_pan'
]
# Iterating through each sieve column to calculate the percentage of total sediment retained on that sieve
for col in sieve_columns:
    # Creating a new column in 'df_perc' with the suffix '_perc'
    # This column stores the percentage of sediment retained on the sieve relative to the measured total sediment
    df_perc[col + '_perc'] = (df[col] / df['measured_total_sediment']) * 100
```

```python
# Save the 'df_perc' DataFrame to a CSV file named 'seive_percentage.csv'
# index=False ensures that the row index is not written to the CSV file
df_perc.to_csv('seive_percentage.csv', index=False)
# Print a message to confirm that the file was saved successfully
print("CSV file saved successfully!")
```
