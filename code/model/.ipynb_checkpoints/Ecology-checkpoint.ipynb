```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers, Model
```

```python
def load_and_merge(file_abundance, file_toc, file_grain):
    df1 = pd.read_csv(file_abundance)
    df2 = pd.read_csv(file_toc)
    df3 = pd.read_csv(file_grain)

    df = df1.merge(df2, on=["year_month", "year", "month", "site", "replicate"])
    df = df.merge(df3, on=["year_month", "year", "month", "site", "replicate"])

    return df
```

```python
TIMESTEPS = 12
MISSING_VALUE = 0.0
```

```python
def add_time_features(df):
    df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
    df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)
    return df
```

```python
def prepare_tensor(df, feature_cols):
    groups = df.groupby(["site", "replicate", "year"])
    
    X = []
    for _, g in groups:
        g = g.sort_values("month")
        if len(g) == 12:
            X.append(g[feature_cols].values)
    
    return np.array(X)
```

```python
def create_mask(X):
    return (X != MISSING_VALUE).astype(np.float32)

def random_mask(X, missing_rate=0.2):
    mask = (np.random.rand(*X.shape) > missing_rate).astype(np.float32)
    return X * mask, mask
```

```python
def build_model(timesteps, n_features):
    
    inputs = layers.Input(shape=(timesteps, n_features))
    
    x = layers.Masking(mask_value=MISSING_VALUE)(inputs)
    
    x = layers.Bidirectional(layers.LSTM(128, return_sequences=True))(x)
    x = layers.Bidirectional(layers.LSTM(64, return_sequences=False))(x)
    
    latent = layers.Dense(32, activation='relu')(x)
    
    x = layers.RepeatVector(timesteps)(latent)
    
    x = layers.Bidirectional(layers.LSTM(64, return_sequences=True))(x)
    x = layers.Bidirectional(layers.LSTM(128, return_sequences=True))(x)
    
    outputs = layers.TimeDistributed(layers.Dense(n_features))(x)
    
    model = Model(inputs, outputs)
    model.compile(optimizer='adam', loss='mse')
    
    return model
```

```python
def train_model(model, X, epochs=100, batch_size=16):
    
    X_corrupt, _ = random_mask(X)
    
    model.fit(
        X_corrupt,
        X,
        epochs=epochs,
        batch_size=batch_size,
        validation_split=0.1,
        verbose=1
    )
```

```python
def impute(model, X):
    mask = create_mask(X)
    X_pred = model.predict(X)
    X_filled = mask * X + (1 - mask) * X_pred
    return X_filled
```

```python
# =========================
# MAIN PIPELINE
# =========================
def run_pipeline(file_abundance, file_toc, file_grain):
    
    df = load_and_merge(file_abundance, file_toc , file_grain)
    df = add_time_features(df)
    
    grain_cols = [col for col in df.columns if "grain" in col.lower()]
    
    feature_cols = [
        "Abundance", "Biomass",
        "TOC", "POC",
        *grain_cols,
        "month_sin", "month_cos"
    ]
    
    df[feature_cols] = df[feature_cols].fillna(MISSING_VALUE)
    
    X = prepare_tensor(df, feature_cols)
    
    model = build_model(TIMESTEPS, X.shape[-1])
    
    train_model(model, X)
    
    X_filled = impute(model, X)
    
    return model, X_filled

# =========================
# USAGE
# =========================
# model, X_filled = run_pipeline(
#     "abundance_biomass.csv",
#     "toc_poc.csv",
#     "grain_size.csv"
# )
```

```python
def encoder(hp, timesteps, n_features):

    inputs = layers.Input(shape=(timesteps, n_features))
    
    # Masking layer (important for missing values)
    x = layers.Masking(mask_value=0.0)(inputs)
    
    # LSTM Encoder
    x = layers.LSTM(
        units=hp.Int('enc_units_1', 64, 256, step=64),
        return_sequences=True
    )(x)
    
    x = layers.LSTM(
        units=hp.Int('enc_units_2', 32, 128, step=32),
        return_sequences=False
    )(x)
    
    latent = layers.Dense(
        hp.Int('latent_dim', 16, 64, step=16),
        activation='relu'
    )(x)
    
    return Model(inputs, latent, name="encoder")
```

```python
def decoder(hp, timesteps, n_features):
    
    latent_inputs = layers.Input(shape=(hp.get('latent_dim'),))
    
    x = layers.RepeatVector(timesteps)(latent_inputs)
    
    x = layers.LSTM(
        units=hp.Int('dec_units_1', 32, 128, step=32),
        return_sequences=True
    )(x)
    
    x = layers.LSTM(
        units=hp.Int('dec_units_2', 64, 256, step=64),
        return_sequences=True
    )(x)
    
    outputs = layers.TimeDistributed(
        layers.Dense(n_features)
    )(x)
    
    return Model(latent_inputs, outputs, name="decoder")
```

```python
def build_autoencoder(hp, timesteps, n_features):
    
    enc = encoder(hp, timesteps, n_features)
    dec = decoder(hp, timesteps, n_features)
    
    inputs = enc.input
    latent = enc(inputs)
    outputs = dec(latent)
    
    model = Model(inputs, outputs)
    
    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            hp.Float('lr', 1e-4, 1e-2, sampling='log')
        ),
        loss='mse'
    )
    
    return model
```
