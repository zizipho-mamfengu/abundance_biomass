```python
import matplotlib.pyplot as plt
import numpy as np

# Data
estuaries = [
    "Kasouga (EC)",
    "Great Brak (WC)",
    "Langebaan (WC)",
    "Swartkops (EC)",
    "Durban Bay (KZN)"
]

# maximum abundances
max_abundance = [156, 419, 600, 400, 264]

x = np.arange(len(estuaries))

# Plot
fig, ax = plt.subplots(dpi=500)

# Plot abundance as bars (max in light blue, min in navy if >0)
for i in range(len(estuaries)):
    ax.bar(estuaries[i], max_abundance[i], color="skyblue")
    # Label max values above bars
    ax.text(i, max_abundance[i] * 1.0, str(max_abundance[i]), 
            ha="center", va="bottom", rotation=0)

# Log scale on y-axis
#ax.set_yscale("log")
ax.set_ylabel("Abundance (Individuals/m², log scale)")
ax.set_title("Known sand prawn abundances in SA estuaries")

plt.xticks(rotation=30, ha="right")

plt.tight_layout()
plt.savefig(r"C:\Users\LENOVO\Documents\PostGrad\BSc Hons\4ZOL509_Research\fig\kownstudiesplot.png", dpi=500)
plt.show()
```
